Povoamento de Dados- Dimensão Tipo de Aula

Importação de Pacotes

```python
import pandas as pd
```

Leitura das Fontes de Dados

```python
atividades = pd.read_csv('../Fontes/atividades_ginasio_fonte.csv', encoding='latin1')
print(f"Registos de atividades lidos: {atividades.shape[0]}")
atividades.head()
```

Filtração das Atividades por apenas Aulas de Grupo

```python
aulas = atividades[atividades['tipo_atividade'].str.strip().str.lower() == 'aula de grupo'].copy()
print(f"Aulas identificadas: {aulas.shape[0]}")
```

Verificação da Integridade dos Dados

```python
assert aulas['id_atividade'].is_unique, "IDs de treino não são únicos!"
assert aulas[['nome_atividade', 'duracao_minutos', 'calorias_medias_queimadas', 'objetivo_treino', 'nivel_intensidade']].isnull().sum().sum() == 0, "Existem valores críticos em falta!" 
```

Adicionar elemento de Capacidade

```python
aulas["capacidade"] = 40
```

Seleção dos Atributos

```python
dim_aulas = aulas[[
    'id_atividade', 'nome_atividade', 'duracao_minutos', 'calorias_medias_queimadas',
    'nivel_intensidade', 'capacidade'
]].copy()

dim_aulas = dim_aulas.rename(columns={
    'id_atividade': 'aula_id',
    'nome_atividade': 'aula_nome',
    'duracao_minutos': 'aula_duracao_min',
    'calorias_medias_queimadas': 'aula_calorias', 
    'nivel_intensidade': 'aula_intensidade',
    'capacidade': 'aula_capacidade',
})
```

Gerar Surrogate Key

```python
dim_aulas.insert(1, 'aula_sk', range(1, 1 + len(dim_aulas)))
```

Exportação dos Dados

```python
dim_aulas.to_csv("../Dados Finais/dim_aula.csv", index=False, encoding="utf-8-sig")
print("Dimensão Tipo De Aula pronta para carga!")
dim_aulas.head()
```
