```python
import pandas as pd
from sqlalchemy import create_engine
from sqlalchemy import text
```

```python
# Ligação à base de dados MySQL
engine = create_engine("mysql+pymysql://root:panelis6969@localhost:3306/easyfit_dw")

dim_instrutor = pd.read_csv("../Dados Finais/dim_instrutor.csv",  encoding="utf-8")
dim_treino = pd.read_csv("../Dados Finais/dim_treino.csv",  encoding="utf-8")
dim_aula = pd.read_csv("../Dados Finais/dim_aula.csv",  encoding="utf-8")
dim_cliente= pd.read_csv("../Dados Finais/dim_cliente.csv",  encoding="utf-8")
dim_campanha= pd.read_csv("../Dados Finais/dim_campanha.csv",  encoding="utf-8")

tf_cliente= pd.read_csv("../Dados Finais/tf_cliente.csv",  encoding="utf-8")
tf_instrutor= pd.read_csv("../Dados Finais/tf_instrutor.csv",  encoding="utf-8")
```

```python

# Apagar dados preservando estrutura e constraints
with engine.begin() as conn:
    conn.execute(text("DELETE FROM tf_instrutor"))
    conn.execute(text("DELETE FROM tf_cliente"))
    conn.execute(text("DELETE FROM dim_instrutor"))
    conn.execute(text("DELETE FROM dim_cliente"))
    conn.execute(text("DELETE FROM dim_campanha"))
    conn.execute(text("DELETE FROM dim_treino"))
    conn.execute(text("DELETE FROM dim_aula"))

# Inserir dados atualizados
dim_cliente.to_sql("dim_cliente", con=engine, if_exists='append', index=False)
dim_instrutor.to_sql("dim_instrutor", con=engine, if_exists='append', index=False)
dim_campanha.to_sql("dim_campanha", con=engine, if_exists='append', index=False)
dim_treino.to_sql("dim_treino", con=engine, if_exists='append', index=False)
dim_aula.to_sql("dim_aula", con=engine, if_exists='append', index=False)
tf_cliente.to_sql("tf_cliente", con=engine, if_exists='append', index=False)
tf_instrutor.to_sql("tf_instrutor", con=engine, if_exists='append', index=False)
```
