Povoamento de Dados- Dimensão Instrutor

Importação de Pacotes

```python
import pandas as pd
```

Leitura de Dados da fonte

```python
instrutores = pd.read_csv('../Fontes/instrutores_ginasio_fonte.csv', encoding='latin1')
print(f"Registos lidos: {instrutores.shape[0]}")
display(instrutores.head())
```

Validação da Integridade

```python
print("Duplicados no ID:", instrutores.duplicated(subset=['id_instrutor']).sum())
print("Valores nulos por coluna:\n", instrutores.isnull().sum())

# Garantir unicidade e integridade mínima
assert instrutores['id_instrutor'].is_unique, "IDs de instrutor não são únicos!"
assert instrutores[['id_instrutor', 'nome', 'area_especializacao']].isnull().sum().sum() == 0, "Existem valores críticos em falta!"
```

Seleção e Transformação dos Campos

```python
dim_instrutor = instrutores[['id_instrutor', 'nome', 'area_especializacao']].copy()

# Renomear colunas para consistência dimensional
dim_instrutor = dim_instrutor.rename(columns={
    'id_instrutor': 'instrutor_id',
    'nome': 'instrutor_nome',
    'area_especializacao': 'instrutor_especializacao'
})
```

Gerar Surrogate Key

```python
dim_instrutor.insert(1, 'instrutor_sk', range(1, 1 + len(dim_instrutor)))
```

Exportação dos Dados

```python
dim_instrutor.to_csv("../Dados Finais/dim_instrutor.csv", index=False, encoding="utf-8-sig")
print("Dimensão Instrutor pronta para carga.")
display(dim_instrutor.head())
```
