Povoamento da Dimensão Campanha


```python
import pandas as pd
```

Carregar dados da fonte

```python
campanhas = pd.read_csv('../Fontes/campanhas.csv', encoding='latin1')
print(f"Registos de campanhas lidos: {campanhas.shape[0]}")
campanhas.head()
```

Verificação de Integridade

```python
# Verificar valores nulos
print("Valores em falta:")
print(campanhas.isnull().sum())

# Verificar duplicados com base na chave natural (ID)
print("\nDuplicados com base no ID:")
print(campanhas.duplicated(subset=["id_campanha"]).sum())
```

Seleção dos Atributos

```python
dim_campanha = campanhas[[
    'id_campanha', 'nome', 'tipo', 'data_inicio',
    'data_fim', 'descricao', 'publico_alvo'
]].copy()

dim_campanha = dim_campanha.rename(columns={
    'id_campanha': 'campanha_id',
    'nome': 'campanha_nome',
    'tipo': 'campanha_tipo',
    'data_inicio': 'campanha_data_inicio', 
    'data_fim': 'campanha_data_fim',
    'descricao': 'campanha_descricao',
    'publico_alvo': 'campanha_publico_alvo'
})
```

Gerar Surrogate Key

```python
dim_campanha.insert(1, 'campanha_sk', range(1, 1 + len(dim_campanha)))
```

Exportação dos Dados

```python
dim_campanha.to_csv("../Dados Finais/dim_campanha.csv", index=False, encoding="utf-8-sig")
print("Dimensão Campanha pronta para carga!")
dim_campanha.head()
```
